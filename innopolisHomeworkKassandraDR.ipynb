```python
import random
import statistics
```

```python
def random_list(a: int, b: int, length: int) -> list:
    return [random.randint(a, b) for _ in range(length)]
```

```python
numbers = random_list(1, 10, 20)

print(numbers)
```

```python
def frequency_dict(values: list) -> dict:
    result = {}

    for value in values:
        if value in result:
            result[value] += 1
        else:
            result[value] = 1

    return result
```

```python
test_list = [1, 2, 3, 1, 2, 2]

print(frequency_dict(test_list))
```

```python
def statistics_list(values: list) -> dict:
    return {
        "минимум": min(values),
        "среднее": statistics.mean(values),
        "медиана": statistics.median(values),
        "мода": statistics.mode(values),
        "максимум": max(values)
    }
```

```python
print(statistics_list(numbers))
```

```python
numbers = random_list(1, 10, 20)

print("Сгенерированный список:")
print(numbers)

print("\nЧастотный ряд:")
print(frequency_dict(numbers))

print("\nСтатистики:")
print(statistics_list(numbers))
```
